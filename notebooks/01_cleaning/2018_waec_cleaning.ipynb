```python
# Importing the required libraries
import pandas as pd
from google.colab import files
```

# PRIVATE

```python
priv_df = pd.read_csv("/content/2018 Private WAEC Data.csv", header=None)
```

```python
# Viewing the DataFrame
priv_df.head(10)
```

```python
# Renaming the columns
priv_df.columns = [
    "state",
    "cands_male",
    "cands_female",
    "cands_total",
    "creds_5_eng_male",
    "creds_5_eng_female",
    "creds_5_eng_total",
    "creds_5_maths_male",
    "creds_5_maths_female",
    "creds_5_maths_total",
    "creds_5_eng_maths_male",
    "creds_5_eng_maths_female",
    "creds_5_eng_maths_total",
    "creds_5_eng_maths_male_pct",
    "creds_5_eng_maths_female_pct",
    "creds_5_eng_maths_total_pct"
]
```

```python
# Creating a function to handle the column names
def state_handling(df):
  df["state"] = df["state"].astype("string").str.title()

  df["state"] = df["state"].replace(
      {"Abuja": "FCT"}
  )

  nigeria_column = df[df["state"] == "Nigeria"]
  df = df[df["state"] != "Nigeria"]

  df = df.sort_values("state")

  df = pd.concat([df, nigeria_column])

  df = df.reset_index(drop=True)

  return df
```

```python
# Creating a function to handle the numerical columns
def numeric_handling(df, num_cols):
  for col in num_cols:
    df[col] = df[col].astype("string").str.replace(",", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

  return df
```

```python
# Viewing the result
priv_df
```

```python
# Dropping the irrelevant rows
priv_df = priv_df.drop(axis=0, index=[0, 1, 2, 3, 4]).reset_index(drop=True)
```

```python
# Viewing the result
priv_df
```

```python
priv_df = state_handling(priv_df)
```

```python
# Viewing the result
priv_df
```

```python
# Handling the numerical columns
num_cols = [
    "cands_male",
    "cands_female",
    "cands_total",
    "creds_5_eng_male",
    "creds_5_eng_female",
    "creds_5_eng_total",
    "creds_5_maths_male",
    "creds_5_maths_female",
    "creds_5_maths_total",
    "creds_5_eng_maths_male",
    "creds_5_eng_maths_female",
    "creds_5_eng_maths_total",
    "creds_5_eng_maths_male_pct",
    "creds_5_eng_maths_female_pct",
    "creds_5_eng_maths_total_pct"
]
priv_df = numeric_handling(priv_df, num_cols)
```

```python
# Viewing the result
priv_df
```

```python
# Saving the DataFrame
file_name = "Cleaned 2018 Private WAEC Data.csv"
priv_df.to_csv(file_name, index=False)
files.download(file_name)
```
