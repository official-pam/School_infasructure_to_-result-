# 2017 PRIVATE STUDENT DATA

```python
# Loading the dataframe
priv_students_17 = pd.read_csv("/content/2017 Private Student Data.csv")
```

```python
# Viewing the first five rows of the dataframe
priv_students_17.head()
```

```python
# Renaming the columns
priv_students_17.columns = [
    "state",
    "num_schools_and_ss1_male",
    "ss1_female",
    "missing_values",
    "ss2_male",
    "ss2_female",
    "ss3_male",
    "ss3_female",
    "total_male",
    "total_female_and_students"
]
```

```python
# Viewing the result
priv_students_17.head()
```

```python
# Viewing the result (contd.)
priv_students_17.tail()
```

```python
# Dropping the rows at index 0, 1, 2, and 41
priv_students_17 = priv_students_17.drop(index=[0, 1, 2, 41]).reset_index(drop=True)
```

```python
# Viewing the result
priv_students_17.head()
```

```python
# Viewign the result (contd.)
priv_students_17.tail()
```

```python
# Dropping the column filled with missing values
priv_students_17 = priv_students_17.dropna(axis=1, how="all")
```

```python
# Viewing the result
priv_students_17.head()
```

```python
priv_students_17
```

```python
# Replacing the 'ss2_male' values for Lagos, Osun, and National
jumbled_states = ["Lagos", "Ogun", "National"]

for state in jumbled_states:
  mask = priv_students_17["state"] == state

  priv_students_17.loc[mask, "ss2_male"] = (
      priv_students_17.loc[mask, "missing_values"]
  )
  priv_students_17.loc[mask, "missing_values"] = np.nan
```

```python
# Viewing the result
priv_students_17
```

```python
# Dropping the column filled with missing values
priv_students_17 = priv_students_17.dropna(axis=1, how="all")
```

```python
# Viewing the result
priv_students_17.head()
```

```python
# Separating the aggregated columns
priv_students_17[["num_schools", "ss1_male"]] = (
    priv_students_17['num_schools_and_ss1_male']
    .str
    .split(expand=True)
)
priv_students_17[["total_female", "total_students"]] = (
    priv_students_17["total_female_and_students"]
    .str
    .split(expand=True)
)

# Dropping the aggregated columns
priv_students_17 = (
    priv_students_17.drop(
        columns=["num_schools_and_ss1_male", "total_female_and_students"]
    )
)
```

```python
# Viewing the result
priv_students_17.head()
```

```python
# Reordering the columns
priv_students_17 = (
    priv_students_17[[
    "state",
    "num_schools",
    "ss1_male",
    "ss1_female",
    "ss2_male",
    "ss2_female",
    "ss3_male",
    "ss3_female",
    "total_male",
    "total_female",
    "total_students"
  ]]
)
```

```python
# Viewing the result
priv_students_17.head()
```

```python
# Viewing the data types of the columns
priv_students_17.dtypes
```

```python
# Converting the numerical columns
num_cols = [
  "num_schools",
  "ss1_male",
  "ss1_female",
  "ss2_male",
  "ss2_female",
  "ss3_male",
  "ss3_female",
  "total_male",
  "total_female",
  "total_students"
]

# Removing the commas in the numerical columns
for col in num_cols:
  priv_students_17[col] = (
      priv_students_17[col]
      .str
      .replace(",", "", regex=False)
  )

  # Converting them from categorical to numerical columns
  priv_students_17[col] = (
      pd.to_numeric(priv_students_17[col]).astype("Int64")
  )
```

```python
# Viewing the result
priv_students_17.dtypes
```

```python
# Converting the result to a CSV and downloading it
file_name = "Cleaned 2017 Private Student Data.csv"

priv_students_17.to_csv(file_name, index=False)
files.download(file_name)
```
